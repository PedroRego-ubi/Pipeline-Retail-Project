# 00 — Data Exploration

Exploratory notebook — **not part of the pipeline**.

Goals:
- Inspect `events.csv` distributions
- Plot gap histograms per (gender, age_range) bucket to calibrate `MAX_GAP_S`
- Check demographic mix and zone distribution
- Verify event type frequencies

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/events.csv', parse_dates=['timestamp'])
print(df.shape)
df.head()
```

```python
# Event type distribution
df['event_type'].value_counts()
```

```python
# Demographic mix
print(df['gender'].value_counts(normalize=True).round(3))
print(df['age_range'].value_counts(normalize=True).round(3))
```

```python
# Inter-event gap distribution per (gender, age_range) bucket
# Use this to calibrate MAX_GAP_S
df_sorted = df.sort_values(['gender', 'age_range', 'timestamp'])
df_sorted['gap_s'] = df_sorted.groupby(['gender', 'age_range'])['timestamp'].diff().dt.total_seconds()
gaps = df_sorted['gap_s'].dropna()

print('Gap percentiles (seconds):')
print(gaps.quantile([0.5, 0.75, 0.90, 0.95, 0.99]).round(0))
```

```python
# Zone visit distribution
df['zone_id'].value_counts()
```
